```python
import pandas as pd
```

```python
data1 = pd.read_csv('results/political/subk_ece_results_political_1.0_95_prob_mass_abs_linear.csv')
data2 = pd.read_csv('results/political/subk_ece_results_political_1.0_uniform_abs_linear.csv')
data1.drop(columns=['k_label'], inplace=True)
data2.drop(columns=['k_label'], inplace=True)
data1 = data1.groupby(['model','k']).mean().reset_index()
data2 = data2.groupby(['model','k']).mean().reset_index()
# Drop -1 ece values
data1 = data1[data1['ece'] != -1]
data2 = data2[data2['ece'] != -1]
# Build ranking of methods for each k
corr_results_political = []
for k in data1['k'].unique():
    subset1 = data1[data1['k'] == k].sort_values('ece')
    subset2 = data2[data2['k'] == k].sort_values('ece')
    print(f"Ranking for k={k} (top 10):")
    ranking1 = subset1['model'].values
    print(ranking1)
    print(subset1[['model', 'ece']])
    print(f"Ranking for k={k} (uniform):")
    ranking2 = subset2['model'].values
    print(ranking2)
    print(subset2[['model', 'ece']])
    # Compare rankings (e.g., using Kendall's tau)
    from scipy.stats import spearmanr
    correlation, _ = spearmanr(ranking1, ranking2)
    corr_results_political.append({'k': k, 'spearman_correlation': correlation})
```
