```python
from datasets import load_dataset
import numpy as np
import pandas as pd
df = load_dataset("allenai/reward-bench-2")
```

```python
df.column_names
```

```python
test_data = df["test"]
pd_test_data = pd.DataFrame(test_data)
```

```python
creater_models = test_data['models']
creater_models
```

```python
model_list = list(creater_models)
four_model_list = [
    model for model in model_list if len(model) == 4]
array_four_model = np.array(four_model_list)
array_four_model.shape
```

```python
unqiue_four_models = np.unique(array_four_model, axis=0)
unique_not_all_identical_four_models = [
    model for model in unqiue_four_models if not np.all(model == model[0])]
unqiue_four_models = unique_not_all_identical_four_models
count_unique_models = {
    tuple(model): np.sum(np.all(array_four_model == model, axis=1))
    for model in unqiue_four_models
}
max(count_unique_models.values())
```

```python
most_common_model = max(count_unique_models, key=count_unique_models.get)
most_common_model
```

```python
pd_test_data[pd_test_data['models'].apply(lambda x: x == list(most_common_model))]
```

## Investigate groupin the models by name

```python
flattened = pd_test_data.models.values.tolist()
flattened = [item for sublist in flattened for item in sublist]
np.unique(flattened)
```

```python
def merging_model_names(model_name):
    if 'gpt' in model_name.lower():
        return 'Gpt'
    elif 'llama' in model_name.lower():
        return 'Llama'
    elif 'claude' in model_name.lower():
        return 'Claude'
    elif 'qwen' in model_name.lower():
        return 'Qwen'
    elif 'mistral' in model_name.lower():
        return 'Mistral'
    elif 'deepseek' in model_name.lower():
        return 'Deepseek'
    else:
        return model_name.split('-')[0]



pd_test_data.models = pd_test_data.models.apply(lambda x: [merging_model_names(y) for y in x])
```

```python
only_four_models = pd_test_data[pd_test_data['models'].apply(lambda x: len(x) == 4)]
only_four_models.shape
```

```python
only_four_models_with_differnt_creaters = only_four_models[
    only_four_models["models"].apply(lambda x: np.unique(x).shape[0] == 4)
    & (only_four_models["num_correct"] == 1)
]
only_four_models_with_differnt_creaters.shape
```

```python
only_four_models_with_differnt_creaters.columns
```

```python
only_four_models_with_differnt_creaters.loc[:, "models"] = only_four_models_with_differnt_creaters.models.apply(lambda x: tuple(sorted(x)))
only_four_models_with_differnt_creaters.groupby('models').size().sort_values(ascending=False)
```

```python
size_of_four_models_with_different_creaters = only_four_models_with_differnt_creaters.groupby("models").size().sort_values(
    ascending=False
)
size_of_four_models_with_different_creaters = size_of_four_models_with_different_creaters[
    size_of_four_models_with_different_creaters >= 10
]
most_frequent_combinations = list(size_of_four_models_with_different_creaters.index)
most_frequent_combinations
```

```python
result_dict = {}
for combination in most_frequent_combinations:
    row_having_most_frequent_combination = only_four_models_with_differnt_creaters.models.apply(lambda x: sorted(x) == list(combination))
    result_dict['-'.join(combination)] = only_four_models_with_differnt_creaters[row_having_most_frequent_combination].id.apply(lambda x: str(x)).tolist()
```

```python
import json
with open('label_ranking_datasets.json', 'w') as f:
    
    json.dump(result_dict, f)
```
